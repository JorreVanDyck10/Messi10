Checks the invalid lines and tells what is wrong with the label annotation.

```python

from pathlib import Path

label_dir = Path("archive/train/labels")

invalid_files = 0
invalid_lines = 0

for label_file in label_dir.glob("*.txt"):

    with open(label_file, "r") as f:
        lines = f.readlines()

    for line_number, line in enumerate(lines, start=1):

        values = line.strip().split()

        if not values:
            continue

        if len(values) != 5:
            invalid_files += 1
            invalid_lines += 1

            print("\nInvalid annotation:")
            print("File:", label_file.name)
            print("Line:", line_number)
            print("Content:", repr(line))
            print("Number of values:", len(values))

print("\n========== SUMMARY ==========")
print("Invalid annotation lines:", invalid_lines)
print("Invalid annotation files:", invalid_files)
```
