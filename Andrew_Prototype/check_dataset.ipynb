This is a dataset to check if all images and labels match each other. It also checks if all class IDs are valid.

```python
from pathlib import Path
from collections import Counter

DATASET = Path("archive")

for split in ["train", "valid", "test"]:

    image_dir = DATASET / split / "images"
    label_dir = DATASET / split / "labels"

    images = list(image_dir.glob("*"))
    label_files = list(label_dir.glob("*.txt"))

    print(f"\n--- {split.upper()} ---")
    print("Images:", len(images))
    print("Labels:", len(label_files))

    missing_labels = 0
    invalid_classes = 0
    class_counts = Counter()

    for image in images:

        label = label_dir / f"{image.stem}.txt"

        if not label.exists():
            missing_labels += 1
            continue

        with open(label, "r") as f:

            for line in f:

                values = line.strip().split()

                if len(values) != 5:
                    print("Invalid annotation:", label)
                    continue

                class_id = int(values[0])

                if class_id not in range(4):
                    invalid_classes += 1
                else:
                    class_counts[class_id] += 1

    print("Missing labels:", missing_labels)
    print("Invalid classes:", invalid_classes)
    print("Class distribution:", class_counts)
```
